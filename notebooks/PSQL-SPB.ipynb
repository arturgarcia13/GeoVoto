```python
# 1. Agregar votos válidos por município e candidato
votos_validos_municipio = votacao_candidato_municipio_zona.groupby(
    ['FK_Cod_Municipio', 'FK_Num_Candidato']
)['Votos_Validos_Candidato'].sum().reset_index()

# 2. Juntar com status de apoio
df_apoio = votos_validos_municipio.merge(
    apoio_prefeito_candidato,
    how='left',
    on=['FK_Cod_Municipio', 'FK_Num_Candidato']
)

# 3. Preencher nulos com 'indeciso'
df_apoio['Status_Apoio'] = df_apoio['Status_Apoio'].fillna('indeciso')

# 4. Calcular média de votos válidos por status de apoio
medias_validos = df_apoio.groupby('Status_Apoio')['Votos_Validos_Candidato'].mean()
print("Média de votos válidos por status de apoio:")
print(medias_validos)

# 5. Calcular impacto (apoia vs não apoia)
media_apoia = medias_validos.get('apoia', 0)
media_nao_apoia = medias_validos.get('nao apoia', 1)  # evitar divisão por zero
impacto_validos = (media_apoia - media_nao_apoia) / media_nao_apoia

print(f"\nImpacto percentual do apoio do prefeito (votos válidos): {impacto_validos:.2%}")
```

```python
# Filtrar apenas os votos do seu candidato
votos_candidato = votacao_candidato_municipio_zona[
    votacao_candidato_municipio_zona['FK_Num_Candidato'] == 1221
].copy()

# Somar os votos por município do candidato
votos_mun = votos_candidato.groupby('FK_Cod_Municipio')['Votos_Validos_Candidato'].sum().reset_index()

# Juntar com os dados de eleitorado
votos_mun = votos_mun.merge(
    manifestacao_eleitorado_municipio[['FK_Cod_Municipio', 'Eleitores_Aptos']],
    on='FK_Cod_Municipio',
    how='left'
)

# Calcular percentual de votos sobre o eleitorado
votos_mun['Percentual_Votos'] = votos_mun['Votos_Validos_Candidato'] / votos_mun['Eleitores_Aptos']

# Selecionar municípios com eleitorado alto, votação baixa
votos_mun_estrategico = votos_mun[
    (votos_mun['Eleitores_Aptos'] > votos_mun['Eleitores_Aptos'].quantile(0.75)) &
    (votos_mun['Percentual_Votos'] < votos_mun['Percentual_Votos'].quantile(0.25))
].copy()

ranking_prioritario = votos_mun_estrategico.sort_values(
    by='Eleitores_Aptos', ascending=True
).merge(
    municipio[['Cod_IBGE', 'Nome_Municipio']],
    left_on='FK_Cod_Municipio', right_on='Cod_IBGE',
    how='left'
)[['Nome_Municipio', 'Eleitores_Aptos', 'Votos_Validos_Candidato', 'Percentual_Votos']]

ranking_prioritario
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from typing import Dict, List, Tuple
import warnings
warnings.filterwarnings('ignore')

class CampaignAnalyzer:
    def __init__(self, candidate_number: int = 1221):
        self.candidate_number = candidate_number
        self.impact_rates = {
            'apoia': 0.12,      # +12% nos votos
            'indeciso': 0.03,   # +3% nos votos (potencial menor)
            'nao apoia': -0.15  # -15% nos votos (impacto negativo)
        }
    
    def load_data(self, dataframes: Dict[str, pd.DataFrame]) -> None:
        """Carrega todos os DataFrames necessários"""
        self.dfs = dataframes
        print("✅ Dados carregados com sucesso!")
    
    def calculate_political_support_impact(self) -> pd.DataFrame:
        """
        Calcula o impacto do apoio político nos votos do candidato
        """
        print("\n🔍 Calculando impacto do apoio político...")
        
        # Obter votos atuais do candidato por município
        candidate_votes = self.dfs['votacao_candidato_municipio_zona'].groupby('FK_Cod_Municipio').agg({
            'Votos_Validos_Candidato': 'sum',
            'Votos_Nominais_Candidato': 'sum'
        }).reset_index()
        
        # Filtrar apenas o candidato 1221
        candidate_votes = candidate_votes[
            candidate_votes['FK_Cod_Municipio'].isin(
                self.dfs['votacao_candidato_municipio_zona'][
                    self.dfs['votacao_candidato_municipio_zona']['FK_Num_Candidato'] == self.candidate_number
                ]['FK_Cod_Municipio'].unique()
            )
        ]
        
        # Merge com dados de apoio
        support_impact = candidate_votes.merge(
            self.dfs['apoio_prefeito_candidato'], 
            left_on='FK_Cod_Municipio', 
            right_on='FK_Cod_Municipio', 
            how='left'
        )
        
        # Calcular impacto baseado no status de apoio
        support_impact['Impact_Rate'] = support_impact['Status_Apoio'].map(self.impact_rates).fillna(0)
        support_impact['Votos_Com_Impacto'] = (
            support_impact['Votos_Validos_Candidato'] * (1 + support_impact['Impact_Rate'])
        ).round().astype(int)
        support_impact['Ganho_Votos'] = support_impact['Votos_Com_Impacto'] - support_impact['Votos_Validos_Candidato']
        
        # Merge com dados do município
        support_impact = support_impact.merge(
            self.dfs['municipio'], 
            left_on='FK_Cod_Municipio', 
            right_on='Cod_IBGE', 
            how='left'
        )
        
        print(f"✅ Impacto calculado para {len(support_impact)} municípios")
        return support_impact
    
    def identify_strategic_municipalities(self) -> pd.DataFrame:
        """
        Identifica municípios estratégicos baseado em:
        - Alto eleitorado + baixa votação atual
        - Potencial de crescimento
        - Status de apoio
        """
        print("\n🎯 Identificando municípios estratégicos...")
        
        # Calcular impacto do apoio
        support_data = self.calculate_political_support_impact()
        
        # Dados do eleitorado por município
        electoral_data = self.dfs['manifestacao_eleitorado_municipio'].copy()
        
        # Merge dados
        strategic_analysis = support_data.merge(
            electoral_data, 
            left_on='FK_Cod_Municipio', 
            right_on='FK_Cod_Municipio', 
            how='left'
        )
        
        # Calcular métricas estratégicas
        strategic_analysis['Penetracao_Atual'] = (
            strategic_analysis['Votos_Validos_Candidato'] / 
            strategic_analysis['Votos_Validos_Municipio'] * 100
        ).fillna(0)
        
        strategic_analysis['Potencial_Crescimento'] = (
            strategic_analysis['Eleitores_Aptos'] - strategic_analysis['Votos_Validos_Candidato']
        ).fillna(0)
        
        strategic_analysis['Eficiencia_Campanha'] = (
            strategic_analysis['Ganho_Votos'] / strategic_analysis['Eleitores_Aptos'] * 100
        ).fillna(0)
        
        # Score estratégico (0-100)
        strategic_analysis['Score_Estrategico'] = (
            # Peso maior para municípios com muito eleitorado
            (strategic_analysis['Eleitores_Aptos'] / strategic_analysis['Eleitores_Aptos'].max()) * 30 +
            # Peso para baixa penetração atual (mais espaço para crescer)
            ((100 - strategic_analysis['Penetracao_Atual']) / 100) * 25 +
            # Peso para potencial de crescimento
            (strategic_analysis['Potencial_Crescimento'] / strategic_analysis['Potencial_Crescimento'].max()) * 25 +
            # Peso para eficiência da campanha (ganho com apoio)
            (strategic_analysis['Eficiencia_Campanha'] / strategic_analysis['Eficiencia_Campanha'].max()) * 20
        ).fillna(0)
        
        print(f"✅ Análise estratégica concluída para {len(strategic_analysis)} municípios")
        return strategic_analysis
    
    def create_priority_ranking(self, strategy: str = "balanced", top_n: int = 10) -> pd.DataFrame:
        """
        Cria ranking de municípios prioritários para a campanha
        
        Estratégias disponíveis:
        - "balanced": Score atual (padrão)
        - "growth": Prioriza baixa penetração
        - "consolidation": Prioriza alta penetração + apoio
        - "efficiency": Prioriza melhor ROI
        """
        print(f"\n🏆 Criando ranking dos top {top_n} municípios prioritários...")
        print(f"📊 Estratégia: {strategy}")
        
        strategic_data = self.identify_strategic_municipalities()
        
        if strategy == "growth":
            # Priorizar baixa penetração (mais espaço para crescimento)
            strategic_data['Score_Crescimento'] = (
                # Peso maior para baixa penetração
                ((100 - strategic_data['Penetracao_Atual']) / 100) * 40 +
                # Peso para tamanho do eleitorado
                (strategic_data['Eleitores_Aptos'] / strategic_data['Eleitores_Aptos'].max()) * 30 +
                # Peso para apoio político
                (strategic_data['Status_Apoio'].map({'apoia': 1, 'indeciso': 0.5, 'nao apoia': 0})) * 20 +
                # Peso para potencial absoluto
                (strategic_data['Potencial_Crescimento'] / strategic_data['Potencial_Crescimento'].max()) * 10
            ).fillna(0)
            sort_column = 'Score_Crescimento'
            
        elif strategy == "consolidation":
            # Priorizar alta penetração + apoio
            strategic_data['Score_Consolidacao'] = (
                # Peso para alta penetração
                (strategic_data['Penetracao_Atual'] / 100) * 35 +
                # Peso forte para apoio político
                (strategic_data['Status_Apoio'].map({'apoia': 1, 'indeciso': 0.3, 'nao apoia': 0})) * 35 +
                # Peso para eleitorado
                (strategic_data['Eleitores_Aptos'] / strategic_data['Eleitores_Aptos'].max()) * 30
            ).fillna(0)
            sort_column = 'Score_Consolidacao'
            
        elif strategy == "efficiency":
            # Priorizar melhor ROI (ganho/eleitorado)
            strategic_data['Score_Eficiencia'] = (
                # Peso principal para eficiência
                (strategic_data['Eficiencia_Campanha'] / strategic_data['Eficiencia_Campanha'].max()) * 50 +
                # Peso para apoio político
                (strategic_data['Status_Apoio'].map({'apoia': 1, 'indeciso': 0.5, 'nao apoia': 0})) * 30 +
                # Peso menor para tamanho
                (strategic_data['Eleitores_Aptos'] / strategic_data['Eleitores_Aptos'].max()) * 20
            ).fillna(0)
            sort_column = 'Score_Eficiencia'
            
        else:  # balanced (padrão)
            sort_column = 'Score_Estrategico'
        
        # Ranking por score escolhido
        priority_ranking = strategic_data.sort_values(sort_column, ascending=False).head(top_n)
        
        # Adicionar posição no ranking
        priority_ranking = priority_ranking.reset_index(drop=True)
        priority_ranking['Posicao_Ranking'] = range(1, len(priority_ranking) + 1)
        
        # Selecionar colunas relevantes
        ranking_columns = [
            'Posicao_Ranking', 'Nome_Municipio', 'Status_Apoio', 'Score_Estrategico',
            'Eleitores_Aptos', 'Votos_Validos_Candidato', 'Votos_Com_Impacto', 'Ganho_Votos',
            'Penetracao_Atual', 'Potencial_Crescimento', 'Eficiencia_Campanha'
        ]
        
        # Adicionar a coluna do score usado se não for o padrão
        if strategy != "balanced":
            ranking_columns.insert(3, sort_column)
        
        final_ranking = priority_ranking[ranking_columns].copy()
        
        print(f"✅ Ranking criado com {len(final_ranking)} municípios")
        return final_ranking
    
    def compare_strategies(self, top_n: int = 10) -> Dict:
        """
        Compara diferentes estratégias de priorização
        """
        print("\n🔄 Comparando estratégias de priorização...")
        
        strategies = {
            "balanced": "Estratégia Balanceada (atual)",
            "growth": "Foco em Crescimento (baixa penetração)",
            "consolidation": "Foco em Consolidação (alta penetração + apoio)",
            "efficiency": "Foco em Eficiência (melhor ROI)"
        }
        
        results = {}
        
        for strategy_key, strategy_name in strategies.items():
            ranking = self.create_priority_ranking(strategy=strategy_key, top_n=top_n)
            
            results[strategy_key] = {
                'name': strategy_name,
                'ranking': ranking,
                'metrics': {
                    'total_eleitores': ranking['Eleitores_Aptos'].sum(),
                    'total_votos_atuais': ranking['Votos_Validos_Candidato'].sum(),
                    'total_ganho': ranking['Ganho_Votos'].sum(),
                    'penetracao_media': ranking['Penetracao_Atual'].mean(),
                    'apoio_positivo': (ranking['Status_Apoio'] == 'apoia').sum(),
                    'apoio_negativo': (ranking['Status_Apoio'] == 'nao apoia').sum(),
                    'indecisos': (ranking['Status_Apoio'] == 'indeciso').sum(),
                }
            }
            
            print(f"✅ {strategy_name}: {results[strategy_key]['metrics']['total_ganho']:,} votos de ganho")
        
        return results

    def generate_summary_report(ranking_df):
        """
        Gera relatório resumido da análise
        """
        print("\n📊 Gerando relatório resumido...")
        
        total_eleitores = ranking_df['Eleitores_Aptos'].sum()
        total_votos_atual = ranking_df['Votos_Validos_Candidato'].sum()
        total_votos_potencial = ranking_df['Votos_Com_Impacto'].sum()
        total_ganho = ranking_df['Ganho_Votos'].sum()
        
        report = {
            'resumo_geral': {
                'total_municipios_prioritarios': len(ranking_df),
                'total_eleitores_aptos': total_eleitores,
                'total_votos_atual': total_votos_atual,
                'total_votos_potencial': total_votos_potencial,
                'total_ganho_votos': total_ganho,
                'percentual_ganho': (total_ganho / total_votos_atual * 100) if total_votos_atual > 0 else 0
            },
            'por_status_apoio': ranking_df.groupby('Status_Apoio').agg({
                'Eleitores_Aptos': 'sum',
                'Votos_Validos_Candidato': 'sum',
                'Ganho_Votos': 'sum',
                'Score_Estrategico': 'mean'
            }).round(2).to_dict(),
            'top_5_municipios': ranking_df.head(5)[['Nome_Municipio', 'Score_Estrategico', 'Ganho_Votos']].to_dict('records')
        }
        
        print("✅ Relatório gerado com sucesso!")
        return report
    
    def visualize_analysis(self, ranking_df: pd.DataFrame) -> None:
        """
        Cria visualizações da análise estratégica
        """
        print("\n📈 Criando visualizações...")
        
        fig, axes = plt.subplots(2, 2, figsize=(15, 12))
        fig.suptitle(f'Análise Estratégica - Candidato {self.candidate_number}', fontsize=16, fontweight='bold')
        
        # 1. Score Estratégico vs Ganho de Votos
        ax1 = axes[0, 0]
        scatter = ax1.scatter(ranking_df['Score_Estrategico'], ranking_df['Ganho_Votos'], 
                           c=ranking_df['Eleitores_Aptos'], cmap='viridis', alpha=0.7)
        ax1.set_xlabel('Score Estratégico')
        ax1.set_ylabel('Ganho de Votos')
        ax1.set_title('Score Estratégico vs Ganho de Votos')
        plt.colorbar(scatter, ax=ax1, label='Eleitores Aptos')
        
        # 2. Distribuição por Status de Apoio
        ax2 = axes[0, 1]
        apoio_counts = ranking_df['Status_Apoio'].value_counts()
        ax2.pie(apoio_counts.values, labels=apoio_counts.index, autopct='%1.1f%%', startangle=90)
        ax2.set_title('Distribuição por Status de Apoio')
        
        # 3. Top 10 Municípios por Score
        ax3 = axes[1, 0]
        top_10 = ranking_df.head(10)
        bars = ax3.barh(range(len(top_10)), top_10['Score_Estrategico'])
        ax3.set_yticks(range(len(top_10)))
        ax3.set_yticklabels(top_10['Nome_Municipio'], fontsize=8)
        ax3.set_xlabel('Score Estratégico')
        ax3.set_title('Top 10 Municípios por Score Estratégico')
        ax3.invert_yaxis()
        
        # 4. Penetração Atual vs Potencial de Crescimento
        ax4 = axes[1, 1]
        ax4.scatter(ranking_df['Penetracao_Atual'], ranking_df['Potencial_Crescimento'], 
                   alpha=0.6, c='red')
        ax4.set_xlabel('Penetração Atual (%)')
        ax4.set_ylabel('Potencial de Crescimento')
        ax4.set_title('Penetração vs Potencial de Crescimento')
        
        plt.tight_layout()
        plt.show()
        
        print("✅ Visualizações criadas!")
```
